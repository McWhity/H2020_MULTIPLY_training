## ⏳ Session Timeline: 3-Hour Smart Farming Lab

Below is the roadmap for today's practical session.

### **00:00 - 00:15** | Welcome & Baseline Survey (15 min)
Getting set up, filling out the experience self-assessment, and reviewing the core learning goals.

### **00:15 - 00:45** | Guided Discovery & Student Research (30 min)
An independent exploration phase.

### **00:45 - 01:15** | The Architectural Debrief (30 min)
A collaborative group discussion weighing the advantages and disadvantages of each data delivery method.

### **01:15 - 02:00** | Hands-On Lab Part 1: Programmatic Access (45 min)
Diving into our first Jupyter Notebook workflows. We will write Python code to target an agricultural area of interest (AOI) and query real-world Copernicus, ASF, and GEE APIs.

### **02:00 - 02:45** | Hands-On Lab Part 2: Spectral & Radar Indices (45 min)
Calculating spatial agronomic metrics (NDVI, NDRE, RVI) from scratch. We will analyze the physics of sensor saturation and compare simple mathematical ratios against advanced biophysical models.

### **02:45 - 03:00** | Synthesis, Career Mapping & Wrap-Up (15 min)
Connecting our scripts to commercial production environments. Q&A on how to scale these pipelines for real-world agronomy and enterprise Farm Management Information Systems (FMIS).

#### Discussion Solution Benchmarks

* **Topic 1 Guide (The Download Dilemma & QGIS):** Downloading complete scenes fills local hard drives fast (a single Sentinel-2 tile is ~500MB to 1GB compressed). QGIS is amazing for visual inspection, but it relies on your computer's local RAM/CPU. Streaming data via COGs and STAC APIs shifts the computation to the cloud, allowing you to slice out just the farm boundary pixels and bypassing heavy local hardware limits.
* **Topic 2 Guide (L1C vs. L2A in QGIS):** L1C data contains atmospheric haze and cloud scattering, making a crop look artificially stressed or overly vibrant depending on weather. L2A is "Surface Reflectance" where these artifacts are mathematically removed. If you accidentally download L1C data, you would have to manually run atmospheric correction plugins (like Semi-Automatic Classification Plugin - SCP) inside QGIS before you can trust your calculations.
* **Topic 3 Guide (Choosing Your Weapon):** Global Amazon project = **Google Earth Engine** (optimized for planetary scale and heavy multi-temporal analytics). Local Tomato Fields = **Direct Cloud Datacubes / STAC API** (optimized for lightweight, automated weekly runs). **QGIS** fits perfectly as an *R&D and validation environment* where you test your scripts locally on a few fields before pushing the code to the automated production pipeline.
